```python
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt
```

```python
def reflect_image(image_path, mode):
    img = Image.open(image_path)
    img_array = np.array(img)
    
    if mode == 'horizontal':
        reflected = np.flip(img_array, axis=0)  # y-axis reflection
    elif mode == 'vertical':
        reflected = np.flip(img_array, axis=1)  # x-axis reflection
    elif mode == 'diagonal':
        reflected = np.transpose(img_array, (1, 0, 2))  # y=x reflection
    elif mode == 'anti-diagonal':
        reflected = np.flip(np.flip(np.transpose(img_array, (1, 0, 2)), axis=0), axis=1)

    return reflected

# image 
image_path = '../img/rotation_image.png'
img = Image.open(image_path)
img_array = np.array(img)

# original
plt.figure(figsize=(5, 5))
plt.imshow(np.array(img_array))
plt.axis('off')
plt.show()

# all reflections
modes = ['horizontal', 'vertical', 'diagonal', 'anti-diagonal']
plt.figure(figsize=(12, 8))
row = 2
col = 2

for i, mode in enumerate(modes):
    plt.subplot(2, 2, i+1)
    reflected = reflect_image(image_path=image_path, mode=mode)
    plt.imshow(reflected)
    plt.title(mode.capitalize())
    plt.axis('off')

# Hide empty subplot
plt.axis('off')

plt.tight_layout()
plt.savefig('019_reflection_image.png', dpi=100)
plt.show()
```
