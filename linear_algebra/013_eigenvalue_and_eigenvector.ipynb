```python
# simple matrix
A = np.array([[2, 1],
              [1, 2]])

# eigenvalues and eigenvectors
eigenvalues, eigenvectors = np.linalg.eig(A)

print("Matrix A:")
print(A)
print("\nEigenvalues:", eigenvalues)
print("Eigenvectors:")
print(eigenvectors)

# A * v = λ * v
v1 = eigenvectors[:, 0]  # First eigenvector
lambda1 = eigenvalues[0]  # First eigenvalue

print(f"\nVerification for first eigenvector:")
print(f"A * v1 = {A @ v1}")
print(f"λ1 * v1 = {lambda1 * v1}")
```

### Eigenvalues and the Diagonal: Key Connections

**Eigenvalues are NOT simply the diagonal entries of a matrix**, BUT there are important connections:

| Matrix Type | Relationship |
|-------------|--------------|
| **Diagonal Matrices** | Eigenvalues = diagonal entries ✅ |
| **Triangular Matrices** | Eigenvalues = diagonal entries ✅ |
| **General Matrices** | Eigenvalues are related to the diagonal through the **trace** (sum of diagonal = sum of eigenvalues) |

---

**Key Insight:**

$$
\text{Tr}(A) = \sum_{i} A_{ii} = \sum_{i} \lambda_i
$$

The trace (sum of diagonal entries) equals the sum of eigenvalues — but individual eigenvalues are **not** individual diagonal entries for general matrices!

```python
# Diagonal matrix
D = np.array([[3, 0],
              [0, 5]])

eigvals, eigvecs = np.linalg.eig(D)

print("Diagonal Matrix D:")
print(D)
print(f"\nDiagonal entries: {np.diag(D)}")
print(f"Eigenvalues: {eigvals}")
```

```python
# Upper triangular matrix
T = np.array([[2, 1, 3],
              [0, 4, 2],
              [0, 0, 1]])

eigvals, _ = np.linalg.eig(T)

print("Upper Triangular Matrix T:")
print(T)
print(f"\nDiagonal entries: {np.diag(T)}")
print(f"Eigenvalues: {eigvals}")
```

```python
# General matrix
A = np.array([[2, 1],
              [1, 3]])

eigvals, eigvecs = np.linalg.eig(A)

print("General Matrix A:")
print(A)
print(f"\nDiagonal entries: {np.diag(A)}")
print(f"Eigenvalues: {eigvals}")
print(f"\nSum of diagonal (trace): {np.trace(A)}")
print(f"Sum of eigenvalues: {np.sum(eigvals)}")
print(f"Trace = Sum of eigenvalues: {np.allclose(np.trace(A), np.sum(eigvals))}")

# Product relationship (determinant)
print(f"\nProduct of diagonal: {np.prod(np.diag(A))}")
print(f"Product of eigenvalues (determinant): {np.prod(eigvals)}")
print(f"Product of diagonal ≠ Product of eigenvalues (for general matrices)")
```
