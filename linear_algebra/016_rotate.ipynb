Rotate a point around the origin by a given angle!

- Your point is at position: $p = (3, 1)$
- You want to rotate it **45° counterclockwise** ($\pi/4$ radians)
- Then rotate it **90° clockwise** ($-\pi/2$ radians)



Rotation Matrices

**45° Counterclockwise** ($\theta = \pi/4$):

$$
R_{45} = \begin{bmatrix} \cos(\pi/4) & -\sin(\pi/4) \\ \sin(\pi/4) & \cos(\pi/4) \end{bmatrix}
= \begin{bmatrix} \frac{\sqrt{2}}{2} & -\frac{\sqrt{2}}{2} \\ \frac{\sqrt{2}}{2} & \frac{\sqrt{2}}{2} \end{bmatrix}
$$

**90° Clockwise** ($\theta = -\pi/2$):

$$
R_{-90} = \begin{bmatrix} \cos(-\pi/2) & -\sin(-\pi/2) \\ \sin(-\pi/2) & \cos(-\pi/2) \end{bmatrix}
= \begin{bmatrix} 0 & 1 \\ -1 & 0 \end{bmatrix}
$$

```python

import numpy as np
import matplotlib.pyplot as plt

def rotate_point(p, theta_deg):
    theta = np.radians(theta_deg)
    R = np.array([[np.cos(theta), -np.sin(theta)],
                  [np.sin(theta),  np.cos(theta)]])
    return R @ p

# Your points
p = np.array([3, 1])
p_45 = rotate_point(p, 45)
p_90cw = rotate_point(p, -90)

print('=' * 50)
print('ROTATION RESULTS')
print('=' * 50)
print(f'Original point:   ({p[0]}, {p[1]})')
print(f'45° CCW:          ({p_45[0]:.2f}, {p_45[1]:.2f})')
print(f'90° CW:           ({p_90cw[0]:.2f}, {p_90cw[1]:.2f})')

# Visualization
fig, ax = plt.subplots(figsize=(8, 8))

# Draw rotation path
angles = np.linspace(0, 360, 100)
radius = np.linalg.norm(p)
ax.plot(radius * np.cos(np.radians(angles)), 
        radius * np.sin(np.radians(angles)), 
        'gray', linestyle='--', alpha=0.5, label=f'Radius = {radius:.2f}')

# Plot points
ax.scatter(*p, color='blue', s=200, zorder=5, label='Original')
ax.scatter(*p_45, color='green', s=200, zorder=5, label='45° CCW')
ax.scatter(*p_90cw, color='red', s=200, zorder=5, label='90° CW')

# Draw vectors
for pt, color, label in [(p, 'blue', 'Original'), 
                          (p_45, 'green', '45° CCW'), 
                          (p_90cw, 'red', '90° CW')]:
    ax.arrow(0, 0, pt[0], pt[1], head_width=0.15, head_length=0.15,
             fc=color, ec=color, alpha=0.7)

# Formatting
ax.grid(True, alpha=0.3)
ax.set_xlim(-4, 4)
ax.set_ylim(-4, 4)
ax.set_aspect('equal')
ax.axhline(y=0, color='k', linewidth=0.5)
ax.axvline(x=0, color='k', linewidth=0.5)
ax.set_title('Rotation Game: Car Turning', fontsize=14)
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.legend()

# plt.savefig('/mnt/c/Users/rah/rotation_challenge.png', dpi=150)
print('\n✅ Plot saved as: C:\\Users\\rah\\rotation_challenge.png')
```
