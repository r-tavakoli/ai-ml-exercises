```python
import numpy as np
import matplotlib.pyplot as plt
import string
```

```python
chess_board = np.zeros((8, 8))
for k in np.arange(-8, 9, 2):
    chess_board += np.eye(8, k=k)
chess_board
```

```python
# plot
plt.figure(figsize=(5, 3))
plt.imshow(chess_board, cmap='gray')
plt.xticks(np.arange(0, 8, 1), labels=string.ascii_letters[:8])
plt.yticks(np.arange(0, 8, 1), labels=np.arange(8, 0, -1))
plt.show()
```
