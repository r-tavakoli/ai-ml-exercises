```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
np.random.seed(42) # reproducibility

uniform = np.random.uniform(0, 1, 1000)
normal = np.random.randn(1000)
poisson = np.random.poisson(3, 1000)
binomial = np.random.binomial(10, 0.5, 1000)

plt.figure(figsize=(15, 3))
plt.subplot(141); plt.hist(uniform, bins=30); plt.title('Uniform')
plt.subplot(142); plt.hist(normal, bins=30); plt.title('Normal')
plt.subplot(143); plt.hist(poisson, bins=30); plt.title('Poisson')
plt.subplot(144); plt.hist(binomial, bins=30); plt.title('Binomial')
plt.show()
```

```python
plt.figure(figsize=(7, 3))

gray_img = np.random.uniform(0, 256, size=(100, 100)) # low, high, size
plt.subplot(1, 3, 1)
plt.imshow(gray_img, cmap='gray')
plt.title('Random Uniform')
plt.axis('off')

rgb_img = np.random.poisson(5, size=(100, 100)) # lam, size
plt.subplot(1, 3, 2)
plt.imshow(rgb_img, cmap='gray')
plt.title('Random Poisson')
plt.axis('off')

noise = np.random.normal(0, 1, size=(100, 100)) # mu, sigma, size
plt.subplot(1, 3, 3)
plt.imshow(noise, cmap='gray')
plt.title('Random Normal')
plt.axis('off')
plt.tight_layout()

plt.savefig('009_random_numbers.png', dpi=70)
plt.show()
```
