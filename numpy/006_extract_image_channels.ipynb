```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
```

```python
img = Image.open('../img/rotation_image.png')
img_array = np.array(img)
```

```python
red_channel = img_array[:, :, 0]    # Red
green_channel = img_array[:, :, 1]  # Green
blue_channel = img_array[:, :, 2]   # Blue
alpha_channel = img_array[:, :, 3]  # Alpha (transparency)


plt.figure(figsize=(15, 4))

# first row
plt.subplot(241)
plt.imshow(red_channel)
plt.title('red')

plt.subplot(242) 
plt.imshow(green_channel) 
plt.title('green')

plt.subplot(243) 
plt.imshow(blue_channel)
plt.title('blue')

plt.subplot(244) 
plt.imshow(alpha_channel) 
plt.title('alpha')

# second row
plt.subplot(245)
plt.imshow(red_channel, cmap='Reds')
plt.title('red')

plt.subplot(246) 
plt.imshow(green_channel, cmap='Greens') 
plt.title('green')

plt.subplot(247) 
plt.imshow(blue_channel, cmap='Blues')
plt.title('blue')

plt.subplot(248) 
plt.imshow(alpha_channel, cmap='gray') 
plt.title('alpha')

plt.tight_layout()

plt.savefig('006_extract_image_channels', dpi=100)
plt.show()
```
