```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
```

```python
data = np.arange(24)
print(f"Original: {data}")

# reshape
reshaped = data.reshape(2, 3, 4)
print(f"\nReshaped to (2,3,4):\n{reshaped}")

# transpose
transposed = reshaped.transpose(1, 0, 2)
print(f"\nTransposed:\n{transposed}")

# flatten
flat = transposed.flatten()
print(f"\nFlattened: {flat}")
```

```python
img = Image.open('../img/rotation_image.png')
img_array = np.array(img)

# original
plt.figure(figsize=(15, 4))
plt.subplot(141)
plt.imshow(img_array)
plt.title('original')

# reshape
plt.subplot(142) 
plt.imshow(img_array.reshape(256, 1024, 4)) 
plt.title('reshape')

# transpose
plt.subplot(143) 
plt.imshow(img_array.transpose((1, 0, 2))) # Current axis order: (0, 1, 2) --> (1, 0, 2) swap height and width which makes image rotate
plt.title('transpose')

# flatten
plt.subplot(144) 
plt.imshow(img_array.flatten()[:262144].reshape(512, 512)) 
plt.title('flatten')


plt.savefig('005_reshape_and_transpose', dpi=100)
plt.show()
```
