```python
from PIL import Image
import numpy as np
```

```python
# load image
img = Image.open('../img/rotation_image.png')

# convert to numpy array
img_array = np.array(img)
print(f"Array shape: {img_array.shape}")
print(f"Array dtype: {img_array.dtype}")

invert_colors_array = 255 - img_array

# save image
img_inv = Image.fromarray(invert_colors_array)
img_inv.save('011_load_and_save_image.png')
# img_inv.show()
img_inv
```
