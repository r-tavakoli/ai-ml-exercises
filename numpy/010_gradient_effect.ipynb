```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# gradient
gradient = np.linspace(0, 255, 640*480).reshape(480, 640).astype(np.uint8)

# radial gradient
y, x = np.ogrid[-240:240, -320:320] # returns arrays that are broadcastable
center = np.sqrt(x**2 + y**2)
radial = (center / center.max() * 255).astype(np.uint8)

plt.figure(figsize=(10, 4))

plt.subplot(121)
plt.imshow(gradient, cmap='gray')
plt.title('Gradient')

plt.subplot(122)
plt.imshow(radial, cmap='gray')
plt.title('Radial Gradient')

plt.savefig('010_gradient_effect.png', dpi=70)
plt.show()
```
