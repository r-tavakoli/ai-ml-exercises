```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
```

```python
# gradient image
image = np.linspace(0, 255, 10000).reshape(100, 100).astype(np.uint8)

# plot
plt.figure(figsize=(15, 4))
plt.subplot(141)
plt.imshow(image, cmap='gray')
plt.title('Original')

# slicing
plt.subplot(142) 
plt.imshow(image[50:70, 50:70], cmap='gray') 
plt.title('Cropped')

# fancy indexing
plt.subplot(143) 
plt.imshow(image[::10, ::10], cmap='gray') 
plt.title('Every 10th pixel')

# boolean indexing
mask = image > 128
image_masked = image.copy()
image_masked[~mask] = 0
plt.subplot(144)
plt.imshow(image_masked, cmap='gray')
plt.title('Thresholded')
plt.show()
```

```python
img = Image.open('../img/rotation_image.png')
img_array = np.array(img)

# plot
plt.figure(figsize=(15, 4))
plt.subplot(141)
plt.imshow(img_array, cmap='gray')
plt.title('Original')

# slicing
plt.subplot(142) 
plt.imshow(img_array[0:300, 0:300], cmap='gray') 
plt.title('Cropped')

# fancy indexing
plt.subplot(143) 
plt.imshow(img_array[::10, ::10], cmap='gray') 
plt.title('Every 10th pixel')

# boolean indexing
mask = img_array > 200
image_masked = img_array.copy()
image_masked[~mask] = 0
plt.subplot(144)
plt.imshow(image_masked, cmap='gray')
plt.title('Thresholded')
plt.savefig('004_indexing', dpi=100)
plt.show()
```
