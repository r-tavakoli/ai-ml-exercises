2. Create a new column genres_list that contains only the genre names (as a list of strings) from the parsed genres column.

```python
movies['genres'].iloc[0]
```

```python
def extract_data(values: list, attribute_name: str) -> list:
    output = list()
    for value in values:
        output.append(value[attribute_name])
    return output if len(output)>0 else np.nan

movies['genres_id'] = movies['genres'].apply(extract_data, attribute_name='id')
movies['genres_name'] = movies['genres'].apply(extract_data, attribute_name='name')
```

4. Combine text from overview, genres, keywords into a single "tags" column, and clean it for potential use in a recommendation system.

```python
movies['keyword_names'] = movies['keywords'].apply(extract_data, attribute_name='name')
movies[['keyword_names', 'genres_name', 'overview']].head()
```

```python
import re
def _clean_text(text: str) -> str:
    if not isinstance(text, str):
        return ''
    
    # Convert to lowercase
    text = text.lower()
    
    # Remove punctuation (keep letters, numbers, and spaces)
    text = re.sub(r'[^\w\s]', ' ', text)
    
    # Remove extra spaces
    text = re.sub(r'\s+', ' ', text).strip()
    
    return text

def clean_text(value: list|str) -> list|str:
    output = list()
        
    if isinstance(value, list):
        for v in value:
            output.append(_clean_text(v))
        return output if len(output)>0 else np.nan
    
    if isinstance(value, str):
        return _clean_text(value)
```

```python
columns_to_clean = ['keyword_names', 'genres_name', 'overview']
movies_cleaned_text = pd.DataFrame()
for col in columns_to_clean:
    movies_cleaned_text[col] = movies[col].apply(clean_text)
```

```python
def combine_tags(row):
    tags_parts = []
    
    # Add overview
    if pd.notna(row['overview']):
        tags_parts.append(str(row['overview']))
    
    # Add genres
    if isinstance(row['genres_name'], list) and row['genres_name']:
        tags_parts.append(' '.join(row['genres_name']))
    
    # Add keywords
    if isinstance(row['keyword_names'], list) and row['keyword_names']:
        tags_parts.append(' '.join(row['keyword_names']))
    
    # Combine all parts
    combined = ' '.join(tags_parts)
    
    # Clean the combined text
    return combined
    
movies_cleaned_text['tags'] = movies_cleaned_text.apply(combine_tags, axis=1)
```
